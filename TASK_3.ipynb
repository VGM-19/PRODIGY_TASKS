```python
text = "the cat sat on the mat. the dog ran in the park. the cat chased the dog around the mat. the sun was shining in the park while the dog played."
```

```python
import string

# remove punctuation and lowercase everything
cleaned_text = text.lower()
cleaned_text = cleaned_text.translate(str.maketrans('', '', string.punctuation))

words = cleaned_text.split()
print(words)
```

```python
from collections import defaultdict

transitions = defaultdict(list)

for i in range(len(words) - 1):
    current_word = words[i]
    next_word = words[i + 1]
    transitions[current_word].append(next_word)

print(transitions)
```

```python
import random

def generate_text(transitions, start_word, length=20):
    current_word = start_word
    generated_words = [current_word]

    for _ in range(length - 1):
        next_words = transitions.get(current_word)
        if not next_words:  # dead end, no known next word
            break
        next_word = random.choice(next_words)
        generated_words.append(next_word)
        current_word = next_word

    return ' '.join(generated_words)

result = generate_text(transitions, start_word="the", length=20)
print(result)
```
